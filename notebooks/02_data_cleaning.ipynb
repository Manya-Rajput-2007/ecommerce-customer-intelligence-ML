```python
import pandas as pd

df = pd.read_csv("../data/processed_data01.csv")

df.head()
```

```python
df.duplicated().sum()
```

```python
df['customer_id'].duplicated().sum()
```

```python
df = df.drop_duplicates(subset='customer_id')
```

```python
numeric_columns = [
    'age',
    'annual_income',
    'avg_monthly_spend',
    'purchase_frequency',
    'discount_usage_rate',
    'return_rate',
    'browsing_time_minutes',
    'support_interactions'
]
```

```python
for col in numeric_columns:
    df[col] = df[col].fillna(df[col].mean())
```
