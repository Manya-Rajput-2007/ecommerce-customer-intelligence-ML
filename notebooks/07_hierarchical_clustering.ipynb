```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/db_scaned_data06.csv")


df.head()
```

```python
import pickle

with open("../data/processed_variables.pkl", "rb") as f:
    data = pickle.load(f)

df = data["df"]
features = data["features"]
x_scaled = data["x_scaled"]
x_scaled_df = data["x_scaled_df"]
scaler = data["scaler"]
final_kmeans = ["final_kmeans"]
dbscan_results =["dbscan_results"]
```

```python
from sklearn.cluster import AgglomerativeClustering

hierarchical = AgglomerativeClustering(
    n_clusters=4
)

hierarchical_labels = hierarchical.fit_predict(x_scaled)

df["Hierarchical_Cluster"] = hierarchical_labels
```

```python
from sklearn.cluster import DBSCAN

for eps in [0.3, 0.5, 0.7, 1.0, 1.5, 2.0]:
    
    dbscan = DBSCAN(
        eps=eps,
        min_samples=10
    )
    
    labels = dbscan.fit_predict(x_scaled)
    
    print(f"eps={eps}")
    print(pd.Series(labels).value_counts())
    print("-" * 30)
```

```python
from sklearn.cluster import DBSCAN

dbscan = DBSCAN(
    eps=2.0,
    min_samples=10,
    n_jobs=-1
)

df["DBSCAN_Cluster"] = dbscan.fit_predict(x_scaled)

print(df["DBSCAN_Cluster"].value_counts())
```

```python
from sklearn.metrics import silhouette_score
import pandas as pd

# KMeans
kmeans_score = silhouette_score(
    x_scaled,
    df["Cluster"]
)

# DBSCAN
dbscan_labels = df["DBSCAN_Cluster"]

mask = dbscan_labels != -1

dbscan_score = silhouette_score(
    x_scaled[mask],
    dbscan_labels[mask]
)

# Hierarchical
hierarchical_score = silhouette_score(
    x_scaled,
    df["Hierarchical_Cluster"]
)

comparison = pd.DataFrame({
    "Algorithm": [
        "KMeans",
        "DBSCAN",
        "Hierarchical"
    ],
    
    "Silhouette Score": [
        kmeans_score,
        dbscan_score,
        hierarchical_score
    ]
})

comparison
```
